Feature Encoding

```python
df.select_dtypes(include='object').columns
```

```python
df['Churn'] = df['Churn'].map({'Yes':1,'No':0})
```

```python
df['Churn']
```

```python
binary_col = ['Partner', 'Dependents', 'PhoneService',
       'PaperlessBilling']
```

```python
for col in binary_col:
    df[col] = df[col].map({'Yes':1,'No':0})
```

```python
df['gender'] = df['gender'].map({'Male':1,'Female':0})
```

```python
df
```

```python
df=pd.get_dummies(df,drop_first=True)
```

```python
df.info()
```

```python
df
```

```python
df = df.astype(int)
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('Churn',axis=1)
y = df['Churn']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000,class_weight='balanced')
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import roc_auc_score
y_prob = model.predict_proba(X_test)[:,1]
roc_auc=roc_auc_score(y_test,y_prob)
print('ROC_AUC_ score=',roc_auc)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
```

```python
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',xticklabels=['No','Yes'],yticklabels=["No","Yes"])
plt.xlabel("Predicted")

plt.ylabel("Actual")
plt.show()
```

XGBoost

```python
from xgboost import XGBClassifier
xgb = XGBClassifier(n_estimators =100,max_depth = 5,learning_rate = 0.1, scale_pos_weight= (y_train.value_counts()[0] / y_train.value_counts()[1])
)
```

```python
xgb.fit(X_train,y_train)
```

```python
y_pred_xgb = xgb.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred_xgb))
```

```python
from sklearn.metrics import roc_auc_score
y_prob_xgb = xgb.predict_proba(X_test)[:,1]
print('ROC_AUC_xgb_score',roc_auc_score(y_test,y_prob_xgb))
```
